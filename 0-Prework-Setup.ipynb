## 4. 🌱 Git
**What:** a version control tool that saves "save points" of your project.

**Why:**
- Go back to an older version if something breaks.
- Work in teams without overwriting each other.
- Required to upload projects to **GitHub**.

### 🪟 Windows (Git for Windows)
1. Go to **git-scm.com** → **Download for Windows** and run the installer.
2. Keep the defaults **except** these screens:
   - **Default editor:** choose **Visual Studio Code**.
   - **Initial branch name:** choose **Override** and type **`main`**.
   - **PATH:** keep *Git from the command line and also from 3rd-party software*.
3. Finish. You now have **Git Bash** in the Start menu. Use it for all Git commands.

Or, in PowerShell: `winget install Git.Git`

### 🍎 Mac
Option A (easiest): open **Terminal** and run:
```bash
git --version
```
If Git isn't installed, macOS offers to install the **Command Line Tools**. Click **Install** and wait.

Option B, with Homebrew: `brew install git`

### 🐧 Linux
Git is often already installed. Check with `git --version`. If not:
```bash
sudo apt update && sudo apt install git    # Ubuntu/Debian
sudo dnf install git                       # Fedora
sudo pacman -S git                         # Arch
```

### Set up (once, all systems)
In Git Bash (Windows) or Terminal (Mac/Linux):
```bash
git config --global user.name "Your Name"
git config --global user.email "you@example.com"   # same email as GitHub
git config --global init.defaultBranch main
```
This is just a name tag on your commits. There's no password.

### ✅ Check
`git --version` shows a version number.

## 6. 🔗 Connect Git to GitHub (SSH, once)
GitHub doesn't accept your account password in Git commands, so we use an **SSH key**: your computer keeps the private key, and GitHub gets the public one.

### Step 1: make the key (same on all systems)
In **Git Bash** (🪟) or **Terminal** (🍎 🐧):
```bash
ssh-keygen -t ed25519 -C "you@example.com"   # press Enter at each question
```

### Step 2: copy the public key
| System | Command |
|---|---|
| 🪟 Windows (Git Bash) | `clip < ~/.ssh/id_ed25519.pub` |
| 🍎 Mac | `pbcopy < ~/.ssh/id_ed25519.pub` |
| 🐧 Linux | `cat ~/.ssh/id_ed25519.pub`, then select and copy the output |

The first two copy it straight to your clipboard, with no output.

### Step 3: add it to GitHub
*Settings → SSH and GPG keys → New SSH key*, paste the key, and save.

### Step 4: test it
```bash
ssh -T git@github.com      # type "yes" if asked
```
✅ `Hi your-username! You've successfully authenticated` means it worked.

From now on, always use the **SSH** link of a repo: `git@github.com:user/repo.git`

### 🛟 Common problems
| Problem | System | Fix |
|---|---|---|
| `git` or `code` "not recognized" / "command not found" | All | Close and reopen the terminal, or restart the computer. On Windows, reinstall and tick **Add to PATH** |
| `code` not found | 🍎 | Run the *Shell Command: Install 'code' command in PATH* step again |
| Can't find Git Bash | 🪟 | Start menu → type "Git Bash" |
| Mac asks to install developer tools | 🍎 | Click **Install**. That's how Git gets installed |
| "App can't be opened, unidentified developer" | 🍎 | Right-click the app → **Open** → **Open** |
| `Permission denied` when installing | 🐧 | Put `sudo` before the command |
| `winget` not recognized | 🪟 | Use the normal download from the website instead |
| Slack invite expired | All | Ask the teacher for a new link |
| `Permission denied (publickey)` | All | Redo section 6 and check you copied the **`.pub`** key |

> **Fun fact:** VS Code and the Slack desktop app are both built with **Electron**, which runs on Chromium, the same engine as Chrome. So they're web pages dressed up as apps 😄
